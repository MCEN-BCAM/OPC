# Figure 4 corrected spatial linearity

This notebook records the corrected matched-site spatial normalisation. Every distributed site is simulated alone and the exact sum of those EPSPs defines the linear prediction.

```
from pathlib import Path
import pandas as pd
from IPython.display import Image, display
BASE=Path('.')
```

```
protocol=pd.read_csv(BASE/'figure4_outputs'/'Figure4_protocol_source_data_corrected.csv')
spatial=protocol.query("protocol=='spatial'")
print('Spatial rows:',len(spatial),'Cells:',spatial.cell_id.nunique())
print('Maximum |S(1)-1|:',(spatial.query('n_synapses==1').summation_ratio-1).abs().max())
display(spatial.groupby(['age_group','n_synapses']).summation_ratio.agg(['count','mean','std','median']))
```
